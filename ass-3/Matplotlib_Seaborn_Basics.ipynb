## Import Libraries

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

## Line Plot

**Expected Output: A line graph joining points (1,2), (2,4), (3,6), (4,8), (5,10).**

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y,marker='o')
plt.title('Line Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

## Bar Chart

**Expected Output: Three vertical bars for Math=80, Science=90, English=75.**

```python
subjects=['Math','Science','English']
marks=[80,90,75]
plt.bar(subjects,marks)
plt.title('Bar Chart')
plt.show()
```

## Scatter Plot

**Expected Output: Five scattered points.**

```python
x=[1,2,3,4,5]
y=[5,7,6,8,10]
plt.scatter(x,y)
plt.title('Scatter Plot')
plt.show()
```

## Histogram

**Expected Output: Histogram showing frequency distribution.**

```python
data=[10,20,20,30,30,30,40,40,50]
plt.hist(data,bins=5)
plt.title('Histogram')
plt.show()
```

## Pie Chart

**Expected Output: Pie chart with percentages.**

```python
labels=['A','B','C']
sizes=[40,35,25]
plt.pie(sizes,labels=labels,autopct='%1.1f%%')
plt.title('Pie Chart')
plt.show()
```

## Seaborn Countplot

**Expected Output: Count plot showing number of records for each day.**

```python
tips=sns.load_dataset('tips')
sns.countplot(data=tips,x='day')
plt.show()
```

## Seaborn Scatterplot

**Expected Output: Scatter plot of total_bill vs tip.**

```python
tips=sns.load_dataset('tips')
sns.scatterplot(data=tips,x='total_bill',y='tip')
plt.show()
```

## Seaborn Boxplot

**Expected Output: Box plot comparing total_bill across days.**

```python
tips=sns.load_dataset('tips')
sns.boxplot(data=tips,x='day',y='total_bill')
plt.show()
```
