# Pandas Notebook

## Theory

**Pandas** is a Python library for data analysis and manipulation.

### Main Data Structures
- **Series**: One-dimensional labeled array.

| Index | Value |
|------:|------:|
|0|10|
|1|20|
|2|30|

- **DataFrame**: Two-dimensional table.

| Name | Age | City |
|------|----:|------|
| Ram |20|Kathmandu|
| Hari |22|Pokhara|
| Sita |21|Butwal|

## Import Pandas

```
import pandas as pd
```

## Create Series

```
s = pd.Series([10,20,30,40])
s
```

## Create DataFrame

```
data={'Name':['Ram','Hari','Sita'],'Age':[20,22,21],'City':['Kathmandu','Pokhara','Butwal']}
df=pd.DataFrame(data)
df
```

```
df['Marks']=[80,90,85]
df
```
